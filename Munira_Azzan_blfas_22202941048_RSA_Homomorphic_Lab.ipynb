```python
print("RSA Homomorphic Lab")
print("===================")

# 1. Generate keys
p = 11
q = 13
n = p * q
phi = (p-1) * (q-1)

print(f"p={p}, q={q}")
print(f"n={p}×{q}={n}")
print(f"phi(n)=({p}-1)×({q}-1)={phi}")

# Find e and d
e = 7
for d in range(1, 1000):
    if (e * d) % phi == 1:
        break

print(f"e={e}, d={d}")
print(f"Public: ({e},{n})")
print(f"Private: ({d},{n})")

# 2. Encrypt messages
m1 = 5
m2 = 9

c1 = (m1 ** e) % n
c2 = (m2 ** e) % n

print(f"\nMessages: m1={m1}, m2={m2}")
print(f"Encrypted: E(m1)={c1}, E(m2)={c2}")

# 3. Homomorphic test
c_product = (c1 * c2) % n
decrypted = (c_product ** d) % n
expected = (m1 * m2) % n

print(f"\nHomomorphic Test:")
print(f"E(m1)×E(m2) = {c1}×{c2} mod {n} = {c_product}")
print(f"Decrypt result: {decrypted}")
print(f"Expected (m1×m2): {expected}")

if decrypted == expected:
    print("Homomorphic property works!!!")
else:
    print("Failed")

# 4. Q6 calculation
print("\n--- Question 6 ---")
p, q = 3, 11
n = p * q
phi = (p-1) * (q-1)
e = 7

for d in range(1, 100):
    if (e * d) % phi == 1:
        break

m1, m2 = 2, 5
c1 = (m1 ** e) % n
c2 = (m2 ** e) % n
c_product = (c1 * c2) % n
decrypted = (c_product ** d) % n

print(f"p={p}, q={q}, n={n}")
print(f"e={e}, d={d}")
print(f"E({m1})={c1}, E({m2})={c2}")
print(f"E(m1)×E(m2)={c1}×{c2} mod {n}={c_product}")
print(f"Decrypt: {decrypted}")
print(f"m1×m2={m1}×{m2}={m1*m2}")
print(f" Correct!" if decrypted == m1*m2 else " Wrong")

```
